```python
import pandas as pa

df = pa.read_csv('https://raw.githubusercontent.com/nurfnick/Data_Viz/main/Data_Sets/IOT-temp.csv')

df.head(15).T
```

```python
df['noted_date'] = pa.to_datetime(df['noted_date'], format='%d-%m-%Y %H:%M')
df.info()
```

```python
display(df.head())
```

```python
min_date = df['noted_date'].min()
max_date = df['noted_date'].max()

print(f"Minimum date and time: {min_date}")
print(f"Maximum date and time: {max_date}")
```

Minimum date and time: 2018-07-28 07:06:00
Maximum date and time: 2018-12-08 09:30:00

```python
df_sorted = df.sort_values(by='noted_date')
df_sorted['time_diff'] = df_sorted['noted_date'].diff()
max_interval = df_sorted['time_diff'].max()

print(f"largest interval between temperature checks: {max_interval}")
```

The largest interval between temperature checks is: 11 days 13:38:00



```python
average_datetime = min_date + (max_date - min_date) / 2

print(f"Average date time: {average_datetime}")
```

The average (midpoint) datetime is: 2018-10-02 20:18:00

```python
import pandas as pd

specified_date = pd.to_datetime('09-11-2018', format='%d-%m-%Y')

outside_temps_on_date = df[
    (df['noted_date'].dt.date == specified_date.date()) &
    (df['out/in'] == 'Out')
]

if not outside_temps_on_date.empty:
    average_outside_temp = outside_temps_on_date['temp'].mean()
    print(f"The average outside temperature on {specified_date.strftime('%d-%m-%Y')} was: {average_outside_temp:.2f}°C")
else:
    # Check if there's any data for the specified date at all
    any_data_on_date = df[df['noted_date'].dt.date == specified_date.date()]
    if not any_data_on_date.empty:
        print(f"There is no 'Out' temperature data for {specified_date.strftime('%d-%m-%Y')} in the dataset.")
    else:
        print(f"There is no temperature data for {specified_date.strftime('%d-%m-%Y')} in the dataset.")
```

There is no temperature data for 09-11-2018, however below shows the temperature data for 08-12-2018.
