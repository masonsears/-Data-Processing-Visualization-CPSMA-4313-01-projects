```python
import google_colab_selenium as gs

driver = gs.Chrome()
```

```python
url = 'https://en.wikipedia.org/wiki/Elden_Ring'
```

```python
item_title_selector = 'h1'

click_element_selector = 'a[title="Bandai Namco Entertainment"]'
```

```python
import time
from selenium.webdriver.common.by import By

try:
    print(f"Opening webpage: {url}")
    driver.get(url)
    time.sleep(3)
    print(f"Webpage loaded. Current URL: {driver.current_url}")

    print(f"\nAttempting to extract titles using selector: '{item_title_selector}'")
    item_elements = driver.find_elements(By.CSS_SELECTOR, item_title_selector)
    titles = [element.text.strip() for element in item_elements if element.text.strip()]

    print("\n--- Extracted Titles ---")
    if titles:
        for i, title in enumerate(titles):
            print(f"{i+1}. {title}")
    else:
        print("No titles found with the specified selector. Please check 'item_title_selector'.")
    print("------------------------")

    print(f"\nAttempting to click element with selector: '{click_element_selector}'")
    try:
        click_element = driver.find_element(By.CSS_SELECTOR, click_element_selector)
        click_element.click()
        print("Element clicked successfully.")
        time.sleep(2)
        print(f"Current URL after click: {driver.current_url}")
    except Exception as click_error:
        print(f"Could not click element: {click_error}")
        print("Please ensure the 'click_element_selector' is correct and the element is present and clickable.")

finally:
    driver.quit()
    print("WebDriver closed.")
```
