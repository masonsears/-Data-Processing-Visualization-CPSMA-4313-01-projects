```python
import pandas as pd
```

```python
data = {
    "age": ["20", "21", "19"],
    "gpa": ["3.5", "3.7", "3.9"],
    "enrolled": ["True", "False", "True"]
}
df = pd.DataFrame(data)
```

```python
df['age'] = df['age'].astype(int)
df['gpa'] = df['gpa'].astype(float)
df['enrolled'] = df['enrolled'].map({'True': True, 'False': False})
```

```python
average_gpa = df['gpa'].mean()

print(f"Average GPA: {average_gpa}")
print(f"Data Structure: {type(data)}")
```
