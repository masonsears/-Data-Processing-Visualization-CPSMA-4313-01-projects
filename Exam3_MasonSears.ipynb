```python
df['day_of_week'] = df['activity_day'].dt.day_name()

max_cadence_by_day_workout = df.groupby(['day_of_week', 'workout_type'])['max_cadence'].max()

max_cadence_overall = max_cadence_by_day_workout.idxmax()
max_cadence_value = max_cadence_by_day_workout.max()

display(f"Day and Workout Type: {max_cadence_overall}")
display(f"Max Cadence: {max_cadence_value}")
```
