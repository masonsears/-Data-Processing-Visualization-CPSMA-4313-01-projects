```python
import requests
import pandas as pa
from bs4 import BeautifulSoup

headers = {'User-Agent': 'chrome/146'}
r = requests.get('https://en.wikipedia.org/wiki/Elden_Ring', headers=headers)
html_contents = r.text
html_soup = BeautifulSoup(html_contents,"lxml")
```

```python
tables = html_soup.find_all('table')
len(tables)
```

```python
from io import StringIO
import pandas as pa

df = pa.read_html(StringIO(str(tables)))[3]

pa.set_option('display.max_rows', None)
pa.set_option('display.max_columns', None)

display(df)
```
